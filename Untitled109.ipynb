```python
import csv
import random
import os

random.seed(7)

DECISIONS = ["approve", "escalate", "reject"]

TEMPLATES = {
    "aml": [
        ("Customer received a {amt} wire from a jurisdiction on the enhanced "
         "monitoring list; stated purpose is '{purpose}'.", "aml"),
        ("Account shows {n} rapid transfers just under the $10,000 reporting "
         "threshold over {days} days.", "aml"),
        ("Customer requested a same-day wire of {amt} to a newly added "
         "beneficiary with no prior transaction history.", "aml"),
    ],
    "kyc": [
        ("New business customer's registered address does not match the "
         "address on the submitted utility bill.", "kyc"),
        ("Beneficial ownership documents list a {pct}% owner who could not "
         "be verified against any public registry.", "kyc"),
        ("Customer's stated occupation and stated monthly income of {amt} "
         "are inconsistent with account activity.", "kyc"),
    ],
    "credit": [
        ("Applicant's debt-to-income ratio is {dti}% with {months} months of "
         "on-time payment history on existing accounts.", "credit"),
        ("Applicant has a thin credit file ({months} months of history) but "
         "a stable employment record.", "credit"),
    ],
}

PROMPT_FRAMES = [
    "You are a compliance analyst. Given the case below, decide: approve, "
    "escalate, or reject. Case: {case}",
    "Case file: {case}\nAs an AML/KYC reviewer, what is your decision "
    "(approve / escalate / reject) and why?",
    "Review this alert and choose one action — approve, escalate, or "
    "reject — for the following situation: {case}",
    "{case}\nBased on standard compliance policy, should this be approved, "
    "escalated for manual review, or rejected?",
    "Acting as a risk officer, classify the following case as APPROVE, "
    "ESCALATE, or REJECT: {case}",
]


def make_case(i):
    domain = random.choice(list(TEMPLATES.keys()))
    template, _ = random.choice(TEMPLATES[domain])
    case_text = template.format(
        amt=f"${random.choice([4500, 8200, 9800, 25000, 120000]):,}",
        purpose=random.choice(["consulting fees", "family support",
                                "equipment purchase", "unspecified"]),
        n=random.choice([3, 4, 6, 9]),
        days=random.choice([2, 5, 10]),
        pct=random.choice([15, 22, 30, 51]),
        dti=random.choice([18, 32, 45, 61]),
        months=random.choice([3, 9, 14, 36]),
    )

    # Difficulty drives ground truth AND how much prompt-sensitivity /
    # semantic uncertainty a trustworthy agent should show.
    difficulty = random.choices(
        ["EASY", "AMBIGUOUS", "HARD"], weights=[0.4, 0.35, 0.25]
    )[0]
    if difficulty == "EASY":
        label = random.choices(DECISIONS, weights=[0.75, 0.20, 0.05])[0]
    elif difficulty == "AMBIGUOUS":
        label = random.choices(DECISIONS, weights=[0.35, 0.45, 0.20])[0]
    else:
        label = random.choices(DECISIONS, weights=[0.15, 0.45, 0.40])[0]

    # A cohort of HARD cases is used later to simulate distribution drift
    # (e.g., a new typology the model wasn't validated on).
    drift_cohort = difficulty == "HARD" and random.random() < 0.5

    return {
        "case_id": f"C{i:04d}",
        "domain": domain,
        "case_text": case_text,
        "difficulty": difficulty,
        "ground_truth": label,
        "drift_cohort": drift_cohort,
    }


def main(n_cases=120, out_path="data/financial_decisions.csv"):
    rows = [make_case(i) for i in range(1, n_cases + 1)]
    fieldnames = ["case_id", "domain", "case_text", "difficulty",
                  "ground_truth", "drift_cohort"]

    # Ensure the directory exists before writing the file
    os.makedirs(os.path.dirname(out_path), exist_ok=True)

    with open(out_path, "w", newline="") as f:
        w = csv.DictWriter(f, fieldnames=fieldnames)
        w.writeheader()
        w.writerows(rows)
    print(f"Wrote {len(rows)} cases to {out_path}")


if __name__ == "__main__":
    main()
```

```python
"""
A stand-in for a real LLM call.

Swap `MockAgenticLLM.generate` for a real client (Anthropic, OpenAI, local
model, etc.) that returns (label, confidence). Everything downstream in
this toolkit (semantic uncertainty, calibration, coverage, decision states)
only depends on that interface, so the rest of the pipeline is model-agnostic.

The mock is deliberately imperfect in three specific ways so the metrics
below have something real to detect:
  1. Prompt sensitivity: the same case can flip decisions depending on
     which prompt framing is used, more so on AMBIGUOUS/HARD cases.
  2. Overconfidence: reported confidence runs higher than true accuracy,
     especially on HARD cases (a classic agentic-AI failure mode).
  3. Drift: cases flagged `drift_cohort` are answered as if the model had
     never seen this typology in validation (confidence stays high, but
     accuracy drops) — this is what the revalidation trigger should catch.
"""
import random

DECISIONS = ["approve", "escalate", "reject"]

# Synonym surface forms a real LLM might emit for the same underlying
# decision — used to demonstrate semantic clustering, not literal string match.
SURFACE_FORMS = {
    "approve": ["approve", "approved", "clear to proceed", "no action needed",
                "greenlight the transaction"],
    "escalate": ["escalate", "escalate for manual review", "flag for review",
                 "send to senior analyst", "needs human review"],
    "reject": ["reject", "decline", "block the transaction", "deny",
               "do not proceed"],
}

DIFFICULTY_NOISE = {"EASY": 0.05, "AMBIGUOUS": 0.30, "HARD": 0.55}
DIFFICULTY_BASE_CONF = {"EASY": 0.93, "AMBIGUOUS": 0.80, "HARD": 0.78}


class MockAgenticLLM:
    def __init__(self, seed=13):
        self.rng = random.Random(seed)

    def generate(self, prompt_frame, case_row, temperature=0.7):
        """Returns (surface_label: str, confidence: float in [0,1])."""
        true_label = case_row["ground_truth"]
        difficulty = case_row["difficulty"]
        drift = case_row["drift_cohort"] in (True, "True", "true")

        noise = DIFFICULTY_NOISE[difficulty]
        # prompt framing itself perturbs the effective noise a bit
        frame_bias = (hash(prompt_frame) % 7) / 100.0
        p_correct = max(0.05, 1 - noise - frame_bias)

        if drift:
            # model is confidently wrong more often on unvalidated typologies
            p_correct *= 0.55

        if self.rng.random() < p_correct:
            label = true_label
        else:
            label = self.rng.choice([d for d in DECISIONS if d != true_label])

        # confidence: overconfident baseline, nudged down slightly with
        # temperature, largely insensitive to whether it's actually drifting
        # (that's the point — the model doesn't know it's out of distribution)
        base = DIFFICULTY_BASE_CONF[difficulty]
        conf = base - 0.1 * temperature + self.rng.uniform(-0.05, 0.05)
        conf = min(0.99, max(0.35, conf))

        surface = self.rng.choice(SURFACE_FORMS[label])
        return surface, round(conf, 3)
```

```python
"""
Maps free-text model outputs to canonical decision states and measures
*semantic* uncertainty — disagreement about meaning, not surface wording.

Two model outputs like "escalate for manual review" and "flag for review"
are the same decision state and should NOT count as disagreement. This is
the core move from the workshop's "Defining decision states and semantic
uncertainty" segment: cluster raw outputs into equivalence classes before
measuring uncertainty over them (cf. semantic entropy, Kuhn et al.).

In production, replace `normalize_to_decision_state` with an NLI-based or
embedding-based clustering step for open-ended outputs. Here, since the
task is closed-set (approve/escalate/reject), a keyword mapping is enough
to demonstrate the mechanism honestly.
"""
import math
from collections import Counter

CANONICAL_MAP = {
    "approve": "approve", "approved": "approve",
    "clear to proceed": "approve", "no action needed": "approve",
    "greenlight the transaction": "approve",
    "escalate": "escalate", "escalate for manual review": "escalate",
    "flag for review": "escalate", "send to senior analyst": "escalate",
    "needs human review": "escalate",
    "reject": "reject", "decline": "reject",
    "block the transaction": "reject", "deny": "reject",
    "do not proceed": "reject",
}


def normalize_to_decision_state(surface_text):
    key = surface_text.strip().lower()
    return CANONICAL_MAP.get(key, "UNMAPPED")


def semantic_entropy(surface_outputs):
    """
    Shannon entropy (bits) over the distribution of CANONICAL decision
    states induced by a set of raw outputs for the same case. 0 = every
    sample agreed on meaning; log2(k) = uniform disagreement across k states.
    """
    states = [normalize_to_decision_state(s) for s in surface_outputs]
    counts = Counter(states)
    n = len(states)
    if n == 0:
        return 0.0
    ent = 0.0
    for c in counts.values():
        p = c / n
        ent -= p * math.log2(p)
    return ent


def majority_state(surface_outputs):
    states = [normalize_to_decision_state(s) for s in surface_outputs]
    return Counter(states).most_common(1)[0][0]
```

```python
from collections import Counter

# Moved from src.semantic_uncertainty and hdQQTBncFWXg
CANONICAL_MAP = {
    "approve": "approve", "approved": "approve",
    "clear to proceed": "approve", "no action needed": "approve",
    "greenlight the transaction": "approve",
    "escalate": "escalate", "escalate for manual review": "escalate",
    "flag for review": "escalate", "send to senior analyst": "escalate",
    "needs human review": "escalate",
    "reject": "reject", "decline": "reject",
    "block the transaction": "reject", "deny": "reject",
    "do not proceed": "reject",
}

def normalize_to_decision_state(surface_text):
    key = surface_text.strip().lower()
    return CANONICAL_MAP.get(key, "UNMAPPED")


def stability_score(surface_outputs_by_frame):
    """
    surface_outputs_by_frame: list of raw text outputs, one per prompt frame
    (a single sample per frame is enough to test framing sensitivity;
    combine with semantic_entropy over repeated samples per frame to also
    capture sampling noise).

    Returns the majority-agreement fraction in [0, 1]: 1.0 means every
    prompt framing produced the same decision state.
    """
    states = [normalize_to_decision_state(s) for s in surface_outputs_by_frame]
    if not states:
        return 0.0
    counts = Counter(states)
    majority_count = counts.most_common(1)[0][1]
    return majority_count / len(states)


def flip_pairs(surface_outputs_by_frame, frame_labels=None):
    """Returns list of (frame_i, frame_j) pairs whose decision states disagree."""
    states = [normalize_to_decision_state(s) for s in surface_outputs_by_frame]
    frame_labels = frame_labels or list(range(len(states)))
    flips = []
    for i in range(len(states)):
        for j in range(i + 1, len(states)):
            if states[i] != states[j]:
                flips.append((frame_labels[i], frame_labels[j]))
    return flips
```

```python
"""
Calibration: does stated confidence match empirical accuracy?

A model that says "92% confident" should be right ~92% of the time across
many such cases. Expected Calibration Error (ECE) and Brier score quantify
the gap. This is the statistical backbone the workshop's "Testing prompt
stability, calibration, and coverage" segment builds on.
"""
import numpy as np


def expected_calibration_error(confidences, correctness, n_bins=10):
    """
    confidences: array-like of predicted confidence in [0, 1]
    correctness: array-like of 1/0, whether the (majority) prediction was correct
    Returns (ece, bin_table) where bin_table has per-bin
    (bin_low, bin_high, avg_confidence, avg_accuracy, count).
    """
    confidences = np.asarray(confidences, dtype=float)
    correctness = np.asarray(correctness, dtype=float)
    bin_edges = np.linspace(0, 1, n_bins + 1)
    ece = 0.0
    n = len(confidences)
    bin_table = []
    for i in range(n_bins):
        lo, hi = bin_edges[i], bin_edges[i + 1]
        if i == n_bins - 1:
            mask = (confidences >= lo) & (confidences <= hi)
        else:
            mask = (confidences >= lo) & (confidences < hi)
        count = mask.sum()
        if count == 0:
            bin_table.append((lo, hi, None, None, 0))
            continue
        avg_conf = confidences[mask].mean()
        avg_acc = correctness[mask].mean()
        ece += (count / n) * abs(avg_conf - avg_acc)
        bin_table.append((lo, hi, avg_conf, avg_acc, int(count)))
    return ece, bin_table


def brier_score(confidences, correctness):
    confidences = np.asarray(confidences, dtype=float)
    correctness = np.asarray(correctness, dtype=float)
    return float(np.mean((confidences - correctness) ** 2))
```

```python
"""
Selective prediction: if the system only auto-decides cases above a
confidence threshold t (and routes everything else to a human), what
fraction of cases does it cover, and what's the error rate on the covered
set? This risk-coverage curve is the quantitative basis for setting
proceed/review/abstain thresholds later.
"""
import numpy as np


def risk_coverage_curve(confidences, correctness, thresholds=None):
    confidences = np.asarray(confidences, dtype=float)
    correctness = np.asarray(correctness, dtype=float)
    if thresholds is None:
        thresholds = np.linspace(0.5, 0.99, 25)
    n = len(confidences)
    curve = []
    for t in thresholds:
        mask = confidences >= t
        covered = int(mask.sum())
        coverage = covered / n if n else 0.0
        if covered == 0:
            risk = None
        else:
            risk = 1 - correctness[mask].mean()
        curve.append((float(t), coverage, risk))
    return curve


def coverage_at_risk_budget(confidences, correctness, max_risk=0.05):
    """
    Find the lowest confidence threshold that keeps empirical risk (error
    rate on covered cases) at or below max_risk. Returns (threshold, coverage, risk)
    or None if no threshold achieves the budget.
    """
    curve = risk_coverage_curve(confidences, correctness,
                                 thresholds=np.linspace(0.5, 0.99, 50))
    feasible = [row for row in curve if row[2] is not None and row[2] <= max_risk]
    if not feasible:
        return None
    return min(feasible, key=lambda row: row[0])
```

```python
"""
Runtime governance: turns the diagnostics above into one of four
decision-relevant states, per case, with an auditable reason.

  PROCEED     - safe to let the agent act autonomously
  REVIEW      - route to a human, but not urgent / not a validation failure
  ABSTAIN     - model should not act; signal is too weak or too unstable
  REVALIDATE  - flags a likely benchmark/coverage gap (possible drift),
                distinct from an ordinary low-confidence case: this is a
                signal to the *governance team*, not just the case queue

Thresholds below are illustrative defaults — in a real deployment they'd
be calibrated against the organization's risk appetite using the
coverage/calibration curves (see coverage.py), not hardcoded.
"""
from dataclasses import dataclass

STATES = ["PROCEED", "REVIEW", "ABSTAIN", "REVALIDATE"]


@dataclass
class Thresholds:
    proceed_confidence: float = 0.82
    proceed_semantic_entropy_max: float = 0.55
    proceed_stability_min: float = 0.8
    abstain_confidence: float = 0.55
    revalidate_semantic_entropy_min: float = 1.35  # near-uniform disagreement (log2(3)=1.585 max)


def classify_decision_state(confidence, semantic_entropy, stability_score,
                             thresholds: Thresholds = Thresholds()):
    """
    Returns (state: str, reason: str).
    Order matters: revalidate/abstain checks fire before proceed so that a
    confidently-wrong-looking case (high confidence, high disagreement)
    doesn't get waved through.
    """
    if semantic_entropy >= thresholds.revalidate_semantic_entropy_min:
        return "REVALIDATE", (
            f"near-uniform disagreement across samples "
            f"(semantic entropy={semantic_entropy:.2f}) despite "
            f"confidence={confidence:.2f} — possible coverage gap"
        )

    if confidence < thresholds.abstain_confidence:
        return "ABSTAIN", f"confidence {confidence:.2f} below abstain floor"

    if (confidence >= thresholds.proceed_confidence
            and semantic_entropy <= thresholds.proceed_semantic_entropy_max
            and stability_score >= thresholds.proceed_stability_min):
        return "PROCEED", (
            f"confidence={confidence:.2f}, semantic entropy="
            f"{semantic_entropy:.2f}, prompt stability={stability_score:.2f} "
            f"all within proceed thresholds"
        )

    return "REVIEW", (
        f"does not meet proceed thresholds "
        f"(confidence={confidence:.2f}, semantic entropy="
        f"{semantic_entropy:.2f}, stability={stability_score:.2f}) "
        f"but not severe enough to abstain/revalidate"
    )
```

```python
import csv
import os
import sys

import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt


# The following imports are no longer needed as the functions/classes are already defined in previous cells
# from src.mock_llm import MockAgenticLLM
# from data.generate_dataset import PROMPT_FRAMES
# from src.semantic_uncertainty import semantic_entropy, majority_state
# from src.prompt_stability import stability_score
# from src.calibration import expected_calibration_error, brier_score
# from src.coverage import risk_coverage_curve, coverage_at_risk_budget
# from src.decision_states import classify_decision_state, Thresholds

SAMPLES_PER_FRAME = 3  # repeated samples per prompt frame, for sampling noise
DATA_PATH = "data/financial_decisions.csv"
OUT_DIR = "outputs"


def load_dataset(path=DATA_PATH):
    with open(path, newline="") as f:
        return list(csv.DictReader(f))


def run_case(llm, case_row):
    """Runs every prompt frame x SAMPLES_PER_FRAME on one case, returns a
    dict of per-case diagnostics."""
    all_outputs = []          # every raw output, across frames and samples
    per_frame_first_output = []  # one output per frame, for stability_score
    confidences = []

    for frame in PROMPT_FRAMES:
        for s in range(SAMPLES_PER_FRAME):
            temp = 0.3 if s == 0 else 0.8
            label, conf = llm.generate(frame, case_row, temperature=temp)
            all_outputs.append(label)
            confidences.append(conf)
            if s == 0:
                per_frame_first_output.append(label)

    sem_entropy = semantic_entropy(all_outputs)
    stability = stability_score(per_frame_first_output)
    maj_state = majority_state(all_outputs)
    mean_conf = float(np.mean(confidences))
    correct = int(maj_state == case_row["ground_truth"])

    state, reason = classify_decision_state(mean_conf, sem_entropy, stability)

    return {
        "case_id": case_row["case_id"],
        "domain": case_row["domain"],
        "difficulty": case_row["difficulty"],
        "drift_cohort": case_row["drift_cohort"],
        "ground_truth": case_row["ground_truth"],
        "majority_prediction": maj_state,
        "correct": correct,
        "mean_confidence": round(mean_conf, 3),
        "semantic_entropy": round(sem_entropy, 3),
        "prompt_stability": round(stability, 3),
        "decision_state": state,
        "reason": reason,
    }


def main():
    os.makedirs(OUT_DIR, exist_ok=True)
    dataset = load_dataset()
    llm = MockAgenticLLM(seed=13)

    results = [run_case(llm, row) for row in dataset]

    # ---- write per-case results ----
    results_path = os.path.join(OUT_DIR, "results.csv")
    with open(results_path, "w", newline="") as f:
        w = csv.DictWriter(f, fieldnames=list(results[0].keys()))
        w.writeheader()
        w.writerows(results)

    confidences = [r["mean_confidence"] for r in results]
    correctness = [r["correct"] for r in results]

    # ---- calibration ----
    ece, bin_table = expected_calibration_error(confidences, correctness, n_bins=10)
    brier = brier_score(confidences, correctness)

    # ---- coverage ----
    curve = risk_coverage_curve(confidences, correctness)
    budget_result = coverage_at_risk_budget(confidences, correctness, max_risk=0.05)

    # ---- decision state breakdown ----
    state_counts = {s: 0 for s in ["PROCEED", "REVIEW", "ABSTAIN", "REVALIDATE"]}
    for r in results:
        state_counts[r["decision_state"]] += 1

    proceed_rows = [r for r in results if r["decision_state"] == "PROCEED"]
    proceed_acc = (sum(r["correct"] for r in proceed_rows) / len(proceed_rows)
                   if proceed_rows else None)

    drift_rows = [r for r in results if r["drift_cohort"] in ("True", True)]
    drift_caught = sum(1 for r in drift_rows if r["decision_state"] == "REVALIDATE")

    # ---- plots ----
    _plot_reliability_diagram(bin_table, ece, os.path.join(OUT_DIR, "calibration_plot.png"))
    _plot_risk_coverage(curve, os.path.join(OUT_DIR, "risk_coverage_plot.png"))
    _plot_decision_states(state_counts, os.path.join(OUT_DIR, "decision_states_plot.png"))

    # ---- summary report ----
    report_path = os.path.join(OUT_DIR, "summary.md")
    with open(report_path, "w") as f:
        f.write("# Trustworthy AI Benchmark — Summary\n\n")
        f.write(f"- Cases evaluated: {len(results)}\n")
        f.write(f"- Overall majority-vote accuracy: {np.mean(correctness):.3f}\n")
        f.write(f"- Expected Calibration Error (ECE, 10 bins): {ece:.3f}\n")
        f.write(f"- Brier score: {brier:.3f}\n\n")
        f.write("## Decision state distribution\n\n")
        for s, c in state_counts.items():
            f.write(f"- {s}: {c} ({c/len(results):.1%})\n")
        f.write(f"\n- Accuracy within PROCEED-state cases: {proceed_acc:.3f}\n"
                if proceed_acc is not None else "\n- No cases reached PROCEED.\n")
        f.write(f"\n## Drift detection\n\n")
        f.write(f"- Cases from the (simulated) unvalidated typology cohort: {len(drift_rows)}\n")
        f.write(f"- Of those, flagged REVALIDATE: {drift_caught} "
                f"({drift_caught/len(drift_rows):.1%})\n" if drift_rows else "")
        f.write("\n## Coverage at 5% risk budget\n\n")
        if budget_result:
            t, cov, risk = budget_result
            f.write(f"- Threshold: {t:.2f} | Coverage: {cov:.1%} | Empirical risk: {risk:.1%}\n")
        else:
            f.write("- No threshold in range achieves a 5% risk budget with this (mock) model.\n")

    print(f"Wrote: {results_path}")
    print(f"Wrote: {report_path}")
    print(f"Wrote: {OUT_DIR}/calibration_plot.png, risk_coverage_plot.png, decision_states_plot.png")
    print("\n--- quick summary ---")
    print(f"Accuracy: {np.mean(correctness):.3f}  |  ECE: {ece:.3f}  |  Brier: {brier:.3f}")
    print(f"Decision states: {state_counts}")


def _plot_reliability_diagram(bin_table, ece, path):
    confs = [b[2] for b in bin_table if b[2] is not None]
    accs = [b[3] for b in bin_table if b[3] is not None]
    fig, ax = plt.subplots(figsize=(5, 5))
    ax.plot([0, 1], [0, 1], linestyle="--", color="gray", label="perfect calibration")
    ax.plot(confs, accs, marker="o", color="#2b6cb0", label="observed")
    ax.set_xlabel("Mean predicted confidence (bin)")
    ax.set_ylabel("Empirical accuracy (bin)")
    ax.set_title(f"Reliability diagram (ECE = {ece:.3f})")
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.legend()
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    plt.close(fig)


def _plot_risk_coverage(curve, path):
    thresholds = [c[0] for c in curve]
    coverage = [c[1] for c in curve]
    risk = [c[2] if c[2] is not None else np.nan for c in curve]
    fig, ax1 = plt.subplots(figsize=(6, 4))
    ax1.plot(thresholds, coverage, color="#2b6cb0", label="coverage")
    ax1.set_xlabel("Confidence threshold")
    ax1.set_ylabel("Coverage (fraction auto-decided)", color="#2b6cb0")
    ax2 = ax1.twinx()
    ax2.plot(thresholds, risk, color="#c53030", label="risk (error rate)")
    ax2.set_ylabel("Empirical risk on covered cases", color="#c53030")
    ax1.set_title("Risk–coverage curve")
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    plt.close(fig)


def _plot_decision_states(state_counts, path):
    labels = list(state_counts.keys())
    values = [state_counts[k] for k in labels]
    colors = {"PROCEED": "#2f855a", "REVIEW": "#d69e2e",
              "ABSTAIN": "#718096", "REVALIDATE": "#c53030"}
    fig, ax = plt.subplots(figsize=(5, 4))
    ax.bar(labels, values, color=[colors[l] for l in labels])
    ax.set_ylabel("Number of cases")
    ax.set_title("Runtime governance: decision state distribution")
    for i, v in enumerate(values):
        ax.text(i, v + 0.5, str(v), ha="center")
    fig.tight_layout()
    fig.savefig(path, dpi=150)
    plt.close(fig)


if __name__ == "__main__":
    main()
```
